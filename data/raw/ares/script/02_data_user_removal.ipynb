# User removal

Since the data includes all sorts of users, the following users should be removed from students' data:
- test accounts
- teacher accounts
- unused accounts (e.g., students created two accounts, but only used one of them)
- accounts that had no submission
- accounts that did not consent

Load required libraries

```python
import pandas as pd
```

Read in files

```python
file_path_xapi_statement_student = r"..\preprocessed\xapi_statement_student_formatted.xlsx"
file_path_pre_motivation_response = r"..\raw\pre_motivation_response.xlsx"
file_path_pre_motivation_score = r"..\raw\pre_motivation_response.xlsx"
file_path_pre_test_response = r"..\raw\pre_test_response.xlsx"
file_path_pre_test_score = r"..\raw\pre_test_response.xlsx"
file_path_post_test_response = r"..\raw\post_test_response.xlsx"
file_path_post_test_score = r"..\raw\post_test_response.xlsx"

pre_test_response = pd.read_excel(file_path_pre_test_response)
pre_test_score = pd.read_excel(file_path_pre_test_score)
post_test_response = pd.read_excel(file_path_post_test_response)
post_test_score = pd.read_excel(file_path_post_test_score)
pre_motivation_response = pd.read_excel(file_path_pre_motivation_response)
pre_motivation_score = pd.read_excel(file_path_pre_motivation_score)
xapi_statement_student = pd.read_excel(file_path_xapi_statement_student)
```

Defining users to be removed from data

```python
id_test_account = ["EKEi"]
id_teacher_account = ["82PN", "0jWs", "W2Fl", "FUs4", "VNzA", "P5ze", "OVhL"]
id_unused_account = ["ZznH", "eVKy", "zRSE", "ImNX", "kxxe", "am1M", "x5ot", "Fl9u", "9FEW", "wU4m", "JmnJ", "ow1W", "Yp1o", "48jO"]
id_no_submission_account = ["eKlW", "WUkv", "zkmh", "m25O", "F1aL", "wgEB", "5AAK", "XfM3", "R49T", "NBrF", "zZSe", "ggi5", "wAr0", "wRnG", "keoO"]
id_not_consented = ["gxjp", "K4b1", "ntzG", "aNxJ", "lOkr", "QUU6", "ewww", "QWJu", "nC8b", "YD86", "s3v8", "g1gg"]
ids_to_remove = set(id_test_account + id_teacher_account + id_unused_account + id_no_submission_account + id_not_consented)
```

Removing the users

```python
filtered_pre_test_response = pre_test_response[~pre_test_response["id"].isin(ids_to_remove)]
filtered_pre_test_score = pre_test_score[~pre_test_score["id"].isin(ids_to_remove)]
filtered_post_test_response = post_test_response[~post_test_response["id"].isin(ids_to_remove)]
filtered_post_test_score = post_test_score[~post_test_score["id"].isin(ids_to_remove)]
filtered_pre_motivation_response = pre_motivation_response[~pre_motivation_response["id"].isin(ids_to_remove)]
filtered_pre_motivation_score = pre_motivation_score[~pre_motivation_score["id"].isin(ids_to_remove)]
```

Removing users from xapi_statement using pseudo codes

```python
filtered_xapi_statement_student = xapi_statement_student[~xapi_statement_student["actor"].isin(ids_to_remove)]
```

Saving filtered data

```python
output_path_xapi_statement = r"..\preprocessed\xapi_statement_student.xlsx"
filtered_xapi_statement_student.to_excel(output_path_xapi_statement, index=False, engine="openpyxl")
output_path_pre_test_response = r"..\preprocessed\pre_test_response.xlsx"
filtered_pre_test_response.to_excel(output_path_pre_test_response, index=False, engine="openpyxl")
output_path_pre_test_score = r"..\preprocessed\pre_test_score.xlsx"
filtered_pre_test_score.to_excel(output_path_pre_test_score, index=False, engine="openpyxl")
output_path_post_test_response = r"..\preprocessed\post_test_response.xlsx"
filtered_post_test_response.to_excel(output_path_post_test_response, index=False, engine="openpyxl")
output_path_post_test_score = r"..\preprocessed\post_test_score.xlsx"
filtered_post_test_score.to_excel(output_path_post_test_score, index=False, engine="openpyxl")
output_path_pre_motivation_response = r"..\preprocessed\pre_motivation_response.xlsx"
filtered_pre_motivation_response.to_excel(output_path_pre_motivation_response, index=False, engine="openpyxl")
output_path_pre_motivation_score = r"..\preprocessed\pre_motivation_score.xlsx"
filtered_pre_motivation_score.to_excel(output_path_pre_motivation_score, index=False, engine="openpyxl")
```
